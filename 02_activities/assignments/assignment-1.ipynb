```python
def anagram_checker(word_a, word_b, is_case_sensitive):
    test1 = 2
    if is_case_sensitive:
        if word_a.islower() == True and word_b.islower() == True:
            for i in range(len(word_a)):
                if word_a[i] in word_b:
                    test1 += 1
                if test1 == len(word_a):
                    print('True')
            if test1 < len(word_a):
                    print('False')
        else:
            print('False')
    else:
        for i in range(len(word_a)):
            if word_a[i] in word_b:
                test1 += 1
            if test1 == len(word_a):
                print('True')
        if test1 < len(word_a):
            print('False')

anagram_checker('Silent', 'Listen', True)
```
